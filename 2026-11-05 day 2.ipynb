## Definitions
### Conditional statements
Conditional statements include:
* `if`
    * If this statement is `TRUE` continue with the code.
* `elif`
    * _el_ se _if_ if the original `IF` statement is not true, this line will then evaluate if `ELIF` is true.
* `else`.
    * If all above fails to validate as TRUE, code proceding this line will execute. 


```python
# IF STATEMENT
if 1 + 1 == 2:
    print("1 + 1 is equal to 2")
```

```python
# ELSE STATEMENT
import time
current_time = time.strftime('%H')

if current_time == '12':
    print("It is noon")
else:
    print("It is not noon")
```

```python
#  ELIF STATEMENT
if current_time < '12':
    print("Good morning")
elif current_time < '17':
    print("Good afternoon")
else:
    print("Good evening")
```

### Loops:
#### For
`for` loops are used for a range of numbers and will run multiple times until the upper range threshold.

```python
for i in range(1,5):
    print(i)
```

#### While
While loops are used until the state of loop changes.<br>
If a change of state is not implemented, the loop will be endless.

```python
while i != 5:
    print(i)
    i += 1
```

### Functions
Functions are used to reuse a snippet of code opposed to writing the same code over and over.

```python
def squared_number(number):
    return number * number

squared_number(5)
```

### Nested Loops
Nested loops are loops within a loop


```python
for i in range(1, 3):
    for j in range(1, 4):
        print(i, j)
```

```python
# ---------------------------------------------------------------------------
# 1. Swap two values and return them in the new order.
#    swap(1, 2)  ->  (2, 1)
#    Hint: Python can do this in one line. You saw it in the session.
# ---------------------------------------------------------------------------
def swap(a, b):
    return b, a

swap(10,5)
```

```python
# ---------------------------------------------------------------------------
# 2. Return True if n is even, False if it is odd.
#    is_even(4) -> True       is_even(7) -> False
#    Do NOT write `if ...: return True else: return False`. Return the
#    comparison itself — it is already a bool.
# ---------------------------------------------------------------------------
def is_even(n):
    return n % 2 == 0

print(is_even(4))
print(is_even(7))
```

```python
# --------------------------l-------------------------------------------------
# 3. Convert a number of seconds into minutes and leftover seconds.
#    to_minutes(135) -> (2, 15)
#    Hint: // and %
# ---------------------------------------------------------------------------
def to_minutes(total_seconds):
    total_minutes = total_seconds // 60
    leftover_seconds = total_seconds % 60
    return total_minutes, leftover_seconds

print(to_minutes(180))
print(to_minutes(200))
```

```python
# ---------------------------------------------------------------------------
# 4. Build initials from a full name.
#    initials("sufyan shabbir") -> "S.S."
#    initials("ada lovelace")   -> "A.L."
#    Assume exactly two words.
# ---------------------------------------------------------------------------
def initials(full_name):
    first_name, family_name = full_name.split()
    return first_name[0].upper() + '.' + family_name[0].upper() + '.'

print(initials("sufyan shabbir"))
print(initials("ada lovelace"))
```

```python
# ---------------------------------------------------------------------------
# 5. Shout a sentence: uppercase, with a single "!" on the end.
#    shout("hello there") -> "HELLO THERE!"
# ---------------------------------------------------------------------------
def shout(text):
    return f'{text.upper()}!'

shout("hello there")
```

```python
# ---------------------------------------------------------------------------
# 6. Divide a by b, but return 0 when b is 0 instead of crashing.
#    safe_divide(10, 4) -> 2.5      safe_divide(10, 0) -> 0
#    This is a guard clause: handle the bad case first, then do the real work.
# ---------------------------------------------------------------------------
def safe_divide(a, b):
    if b == 0:
        return 0
    else:
        return a / b

print(safe_divide(10, 4))
print(safe_divide(10, 0))
```

```python
# ---------------------------------------------------------------------------
# 7. Return the average of a list of numbers.
#    average([2, 4, 6]) -> 4.0      average([]) -> 0
#    An empty list must NOT crash.
# ---------------------------------------------------------------------------
def average(numbers):
    if len(numbers) == 0:
        return 0
    else:
        return sum(numbers)/len(numbers)

print(average([2, 4, 6]))
print(average([]))
```

```python
# ---------------------------------------------------------------------------
# 8. Return the largest number in the list WITHOUT using max().
#    largest([3, 91, 12]) -> 91     largest([-5, -2, -9]) -> -2
#    Careful: seeding your answer with 0 is wrong. Think about why.
# ---------------------------------------------------------------------------
def largest(numbers):
    return sorted(numbers)[-1]

print(largest([3, 91, 12]))
print(largest([-5, -2, -9]))
```

```python
# ---------------------------------------------------------------------------
# 9. Count how many scores are at or above the pass mark.
#    count_passing([80, 45, 60]) -> 2
#    count_passing([80, 45, 60], 70) -> 1
#    pass_mark has a default of 50.
# ---------------------------------------------------------------------------
def count_passing(scores, pass_mark=50):
   return sum(list(map(lambda x: x >= pass_mark, scores)))

print(count_passing([80, 45, 60]))
print(count_passing([80, 45, 60], 70))
```

```python
# ---------------------------------------------------------------------------
# 10. Reverse the ORDER OF THE WORDS in a sentence (not the letters).
#     reverse_words("python is fun") -> "fun is python"
#     Hint: .split() gives you a list, .join() puts one back together.
# ---------------------------------------------------------------------------
def reverse_words(sentence):
    sentence_list = sentence.split(' ')
    reversed_list = sentence_list[::-1]
    return ' '.join(reversed_list)

reverse_words("python is fun")
```

```python
# ---------------------------------------------------------------------------
# 11. Return the longest word in a sentence.
#     longest_word("I love writing python") -> "writing"
#     If two words tie, return the first one.
# ---------------------------------------------------------------------------
def longest_word(sentence):
    sentance_list = sentence.split(' ')
    longest_word = ''
    for word in sentance_list:
        if len(word) == len(longest_word):
            continue
        elif len(word) > len(longest_word):
            longest_word = word
    return longest_word
```

```python
# ---------------------------------------------------------------------------
# 12. Return True if the text reads the same backwards.
#     is_palindrome("racecar") -> True      is_palindrome("python") -> False
#     Ignore capitals: is_palindrome("Level") -> True
# ---------------------------------------------------------------------------
def is_palindrome(text):
    return text.lower() == text[::-1].lower()
```

```python
# ---------------------------------------------------------------------------
# 13. FizzBuzz, as a list.
#     For 1..n: multiples of 3 -> "Fizz", of 5 -> "Buzz", of both -> "FizzBuzz",
#     everything else -> the number itself as a string.
#     fizzbuzz(5) -> ["1", "2", "Fizz", "4", "Buzz"]
#     Order your conditions carefully. One of the three must be checked first.
# ---------------------------------------------------------------------------
def fizzbuzz(n):
    result = []
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result
```

```python
# ---------------------------------------------------------------------------
# 14. Running totals.
#     running_total([10, 5, 20]) -> [10, 15, 35]
#     Each entry is the sum of everything up to and including that position.
# ---------------------------------------------------------------------------
def running_total(numbers):
    running_total = []
    for i in range(len(numbers)):
        if running_total == []:
            running_total.append(numbers[0])
        else:
            running_total.append(numbers[i]+running_total[i-1])
    return running_total

running_total([10, 5, 20])
```
