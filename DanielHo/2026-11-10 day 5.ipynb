## Dependency Injection (DI)
Dependency Injection is a technique that decouples components by providing their dependencies from outside the class instead of letting the class construct them itself.<br>
This achieves Inversion of Control (IoC), meaning the class no longer controls the creation of its dependencies, which improves flexibility, testability, and maintainability of the codebase.

Key benefits:
* Modularity - components are separated into smaller, reusable parts
* Testability - dependencies can be replaced with mocks or stubs for unit testing.
* Maintainability - changes to dependencies require updates only in the injection point, not throughtout the code.
Flexibility - different implementations of a dependency can be swapped without modifying the consuming class.


### Common Methods of Dependency Injection in Python
Constructor Injection - dependencies are passed through the class constructor `__init__`` method.
```py
class Service:
    def do_something(self):
        print("Service is doing something")
class Client:
    def __init__(self, service: Service):
        self.service = service
    def perform_task(self):
        self.service.do_something()
```

Setter Injection - dependencies are set after object creation using a setter method.
```py
class Client:
    def __init__(self):
        self.service = None
    def set_service(self, service: Service):
        self.service = service
    def perform_task(self):
        self.service.do_something()
```

Method Injection - dependencies are passed directly to the method that requires them.
```py
class Client:
    def perform_task(self, service: Service):
        service.do_something()
```

## DB Session Injection
Database sessions are managed through dependency injection, ensuring that each request has its own isolated session.<br>
This approach is crucial for maintaining thread safety and preventing connection leaks.<br>
Each request gets its own session, which is used for all database operations during that request, then closed when the response is sent.<br>
The sessions lifecycle management per request ensures that resources are consistently cleaned up, regardless of whether a request succeeds or fails.<br>
This pattern is essential for building scalable applications that avoid connection leakage and maintain high performance under concurrent loads.
