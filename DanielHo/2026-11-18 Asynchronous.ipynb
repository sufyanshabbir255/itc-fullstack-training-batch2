# Asynchronous Events


````python
import time
import asyncio
import nest_asyncio2 as nest_asyncio
````

````python
def task_sync(name):
    print(f"{name} started @ {time.strftime("%H:%M:%S",time.localtime())}")
    time.sleep(2)
    print(f"{name} finished @ {time.strftime("%H:%M:%S",time.localtime())}")

task_sync("A")
task_sync("B")
````

````python
nest_asyncio.apply()

async def task_async(name):
    print(f"{name} started @ {time.strftime("%H:%M:%S",time.localtime())}")
    await asyncio.sleep(2)
    print(f"{name} finished @ {time.strftime("%H:%M:%S",time.localtime())}")
    
async def main():
    await asyncio.gather(
         task_async("A")
        ,task_async("B")
    )

asyncio.run(main())
````

## 1. Concurrency vs Parallelism
Concurrency - A chef cooking multiple meals at the same time

> Multiple tasks make progress during the same period by sharing resources

Parallelism - Multiple chefs cooking one or more meals at the same time.

> Multiple tasks execute at the same time on different CPU cores


## 2. Asyncio Event Loop
A waiter submits an order to the kitchen and returns to the customer floor to serve others.

When an order is ready, the waiter gets the meal and serves it to the customer.

> The asyncio event loop schedules and manages asynchronous tasks


## 3. async def / await
Normal function:
```py
def get_data():
    return data
```

Async function:
```py
async def get_data():
    return "data"
```

Await:
```py
import asyncio
async def get_data():
    await asyncio.sleep(2)
    return "data"
```
> `async def` defines a co-routine

> `await` pauses execution until another asynchronous operation completes.

## 4. asyncio.gather()
Runs multiple async tasks concurrently.

Without gather: 
```py
await user_service()    #   1 second
await orders_service()  #   1 second
await payment_service() #   1 second
                  # Total   3 seconds
```

With gather:
```py
results = await syncio.gather(
     user_service()
    ,orders_service()
    ,payment_service()
)                 # Total   1 second
```
> Runs multiple co-routines concurrently and waits for all of them to finish.

## 5. aiohttp
```py
import aiohttp

async with aiohttp.ClientSession() as session:
    async with session.get(
        "https://api.example"
    ) as response:
        data = await response.json()
```

> An asynchronous HTTP client/server library used to make non-blocking network requests in Python.
