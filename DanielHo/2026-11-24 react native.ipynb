
## State
State stores data.
```jsx
const [count, setCount] = useState(0);
```

Example
```jsx
import React, { useState } from "react";
import { View, Text, Button } from "react-native";

export default function App() {
    const [count, setCount] = useState(0);

    return (
        <View>
            <Text>{count}</Text>

            <Button
                title="Increment"
                onPress={() => setCount(count + 1)}
            />
        </View>
    );
}
```

## Uses of JSX
### 1. Embedding Expressions
JSX allows embedding JavaScript expressions into the HTML-like syntax, using curly braces:
```jsx
const name = "Jonny"
const greeting = <h1>Hello, {name}!</h1>
```
JSX dynamically inserts the value of the `name` variable  into the rendered output.

### 2. Using Attributes
Attributes are specified like HTML but with differences.<br>
Certain attributes are written in camelCase instead of lowercase syntax used in HTML.
```jsx
const element = <img src="" alt="A description" />;
```
CamelCase for Attribute Names: in JSX, some HTML attributes arewritten in camelCase.

For example, `class` becomes `className`, `for` becomes `htmlFor`, and `style` is an object.<br>
This is because `class` and `for` are reserved words in JavaScript.

### 3. Passing Children
Components or elements can accept children just like HTML elements.<br>
Children are nested elements or content that are passed into a component.<br>
This allows for flexible reusable components.
```jsx
const Welcome = (props) => {
    return <div>{props.children}</div>;
};

const App = () => {
    return (
        <Welcome>
            <h1>Hello, World!</h1>
            <p>Welcome to React.</p>
        </Welcome>
    )
}
```
- **Children as Props** - `Welcome` does not explicitly define the content inside it. Instead, it uses `{props.children}` to render any child elements that are passed between the opening and closing tags of the component when it is used.
- **Flexibility** - `App` passes two child elements, `<h1>` and `<p>`, to `Welcome`. By using `{props.children}`, the `Welcome` component can render any child content, making it reusable with different content each time it's used.

### 4. JSX Represents Objects
JSX is not directly rendered as HTML by React; instead it gets complied into JavaScript objects representing virtual DOM elements.<br>
These objects are later used by React to effeciently update the real DOM.

```jsx
const element = React.createElement(
    "button",
    {
        className: "btn",
        onClick: () => alert("Clicked!"),
    },
    "Click Me"
)
```


This JSX code is converted into a JavaScript object:
```js
{
    type    : 'button'
    ,props  : {
        className   : 'btn'
        ,onClick    : () => alert('Clicked!')
        ,children   : ['Click Me']
    }
}
```

- `type: 'button'` - defines the element type (button).
- `props` - contains attributes like:
    - `className: 'btn'` - for styling
    - `onClick: () => alert('Clicked!')` - event handler for click
    - `children: ['Click Me']` - content inside the button
- React converts JSX into a JavaScript object to efficiently render and manage the UI.
