## Foundation
### Object-Oriented Programming
<u>O</u>bject <u>O</u>riented <u>P</u>rogrammming is the programming paradigm that structures software around objects, units containing both data (attributes) and behaviour (methods), to create modular, reusable, and maintainable code.<br>
It relies on classes as blueprints for objects and uses core principles such as encapsulation, inheritance, polymorphism, and abstraction.



````python
# Define a class called Animal
class Animal:
    def __init__(self, name, family, vertebrate):
        self.name       = name
        self.family     = family
        self.vertebrate = vertebrate
    def get_description(self):
        return f" {self.name} {self.family} {self.vertebrate}"
    def on_the_move(self):
        print(f'The {self.name} is on the move!')

# Create objects (instances) of the Animal class
animal1 = Animal("Leopard", True, "Felidae")
animal2 = Animal("Octopus", False, "Mollusks")

# Access attributes and methods
print(animal1.get_description())
animal2.on_the_move()
````


* __Class__: `Animal` is a blueprint for creating animal objects containing:
    * attributes: `name`, `family`, `vertebrate`
    * methods: `get_description`, `on_the_move`
* __Constructor__: `__init__` initialises each new object with specific values.
* __Object/Instance__: `animal1` and `animal2` are instances of the `Animal` class.
* __Methods__: Functions defined inside a class that operate on the object's data `get_description()` and `on_the_move()`
* __Encapsulation__: Attributes are bundled with methods inside teh class, which is a key OOP principle.


### The Four OOP pillars
#### Abstraction
To abstract, is to hide the implementation details inside something like a prototype or a function.<br>
Abstraction can be done using abstract classes.<br>
These are classes that only provide a blueprint and leave the details to child classes.


````python
from abc import ABC, abstractmethod

class Sport(ABC):
    @abstractmethod
    def play(self):
        pass
class Football(Sport):
    def play(self):
        print("Kick the football")
class Tennis(Sport):
    def play(self):
        print("Volley the tennisball")

sports = [Football(), Tennis()]
for ball in sports:
    ball.play()
````


`Sport` is an abstract class. It says that a sport must have a method to play, but doesn't say how.<br>
The classes `Football` and `Tennis` provide the details.

#### Encapsulation
Encapsulation is the practice of bundling data (attributes) and methods that operate on that data within a class, while restricting direct access to some components to protect the internal state.<br>
* Public: can be used anywhere in the program
* _Protected: this tells other programmers it should not be changed directly.
* __Private: These cannot be accessed directly from oputside the class. 


````python
class BankAccount:
    def __init__(self, account_number, balance):
        self._account_number = account_number       # protected
        self.__balance = balance                    # private
    def deposit(self, amount):
        self.__balance += amount
        print(f"Deposited: {amount}. New balance: {self.__balance}")
    def get_balance(self):
        return self.__balance

account = BankAccount("12349876", 500)
account.deposit(250)
print(account.get_balance())
````

````python
print(account.__balance)
````

Encapsulation makes sure that the balance cannot be directly changed from outside.<br>
Instead, we need to use methods like `deposit()` and `get_balance()`

#### Inheritance
Inheritance allows a class to reuse attributes and methods from another class, promoting code reuse and hierarchaical relationships.
* Single Inheritance: One child class inherits from one parent class.
* Multiple Inheritance: One child inherits from more than one parent class.
* Multilevel Inheritance: A child inherits from a parent, and another child inherits from that child.

````python
class Grandparent:
    def greet(self):
        print("Hello from Grandparent")
class Parent(Grandparent):
    def speak(self):
        print("Hello from Parent")
class Child(Parent):
    def play(self):
        print("Hello from Child")

c = Child()
c.greet()       # Inherited from Grandparent
c.speak()       # Inherited from Parent
c.play()        # Defined in Child
````

 `Parent` inherirts from `Grandparent`<br>
    &emsp; and `Child` inherirts from `Parent`<br>
So `Child` has inherited both methods.


````python
class Employee:
    def work(self):
        print("Employee works")
class Manager(Employee):
    def work(self):
        print("Manager micros")
class ProjectManager(Manager):
    def work(self):
        print("Project Manager plans")

pm = ProjectManager()
pm.work()
````

Python uses the __Method Resolution Order__ to determine which subclass method - `work()` to use.

If we need to access the superclass method from the subclass, we need to use the `super()` function.

````python
class Bird:
    def eat(self):
        print("I can eat")
    def fly(self):
        print("fly with wings")
class Hawk(Bird):
    def eat(self):
        super().eat()
        print("I like to eat critters")

bird_of_pray1 = Hawk()
bird_of_pray1.eat()
````

The `eat()` method of the `Hawk` subclass overrides the same method of the `Bird` superclass.<br>
Inside the `Hawk` class, we used `super().eat()` to call the `eat()` method of the `Bird` superclass, from the `Hawk` subclass.<br>
So when we call the `eat()` method using the `bird_of_pray1` object, both the overridden and the superclass version of the `eat()` method is executed.

#### Polymorphism
Polymorphism refers to the use of a single type entity (method, operator, or object) to represent different types in different scenarios.
* Method overriding: A child class changes behaviour of a method from the parent class.
* Duck typing: If an object behaves like another (e.g. it has the same method), Python treats it the same way.

````python
from math import pi

class Shape:
    def __init__(self, name):
        self.name = name
    def area(self):
        pass
    def fact(self):
        return "I am a 2D shape."
    def __str__(self):
        return self.name
class Square(Shape):
    def __init__(self, length):
        super().__init__("Square")
        self.length = length
    def area(self):
        return self.length**2
    def fact(self):
        return "Squares have each angle equal to 90 degrees."
class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius
    def area(self):
        return pi*self.radius**2

a, b = Square(4), Circle(7)
print(b)
print(b.fact())
print(a.fact())
print(b.area())
````

The method `__str__()` has not been overridden in the child classes, and are used from the parent class.

The Python interpreter automatically recognises that the `fact()` method for `object a` is overridden.<br>
So it uses the one defined in the child class.

The `fact()` method for `object b` was not overridden, therefore the `fact()` from the Parent class `Shape` is used.

````python
class Airplane:
    def fly(self):
        print("fly with fuel")

for object in (Hawk(), Airplane()):
    object.fly()
````

Python does not care about the object's class - only that it has a `fly()` method.

Duck typing is a form of polymorphism where objects do not need to share a parent class - they just need to implement the expected interface.


## Software Design & Best Practices
### KISS Principle
__Keep It Simple Stupid__ is a design philsophy emphasising that systems work best when kept simple.

### DRY Principle
__Don't Repeat Yourself__ is a stated as:
_<p align="center">Every piece of knowledge must have a single, unambiguous, authoraitative representation within a system.</p>_

Why DRY matters:
* Maintainability - centralised logic reduces bugs and inconsistences.
* Consistency - uniform behaviour across modules.
* Faster development - reuse saves time and reduces repeated work.
* Improved readability - less clutter and clearer structure.


### SOLID Principles
The SOLID principles are five object-oriented design guidelines:
* <b>S</b>ingle Responsibility Principle
* <b>O</b>pen/Closed Principle
* <b>L</b>iskov Substitution Principles
* <b>I</b>nterface Segregation Principle
* <b>D</b>ependency Inversion Principle

#### 1. Single Responsibility Principle (SRP)
_<p align="center">A class should have only one reason to change</p>_
I.e. a class should have only one job/purpose within the software system.

Example:
* A baker's sole responsiblilty is baking bread.
* if the baker handles inventory, ordering, customer service and cleaning, it violates SRP since multiple responsibilties are combined.
* Each task is a separate responsibility and combining them reduces the baker's ability to bake.
* To follow SRP, the responsibilties should be given to the different individuals or teams, handling a specific task independently. 

#### 2. Open/Closed Principle (OCP)
_<p align="center">Software Entities (class, modules, functions, etc.) should be open for extension, but closed for modification</p>_

##### Example of violating OCP

````python
class Person:
    def __init__(self, name):
        self.name = name
    def __repr__(self):
        return f"Person(name={self.name})"
class PersonStorage:
    def save_to_database(self, person):
        print(f"Save {person} to database")
    def save_to_json(self, person):
        print(f"Save {person} to JSON file")
````

The class `PersonStorage` saves a `Person` object to different formats.<br>
If you want to save the `Person` object to a different format, you would need to modify `PersonStorage` class, violating the Open/Closed Principle.
##### Refactoring to follow OCP

````python
class PersonStorage:
    @abstractmethod
    def save(self, person):
        pass
class PersonDB(PersonStorage):
    def save(self, person):
        print(f"Save {person} to database")
class PersonJSON(PersonStorage):
    def save(self, person):
        print(f"Save {person} to JSON file")
class PersonXML(PersonStorage):
    def save(self, person):
        print(f"Save {person} to XML file")

person = Person("Peterson")
storage = PersonXML()
storage.save(person)
````

##### Benefits of OCP
* Scalability: new features can be added without altering existing code.
* Maintainability: reduces the risk of introducing bugs when extending functionality.
* Flexibility: Encourages the use of polymorphism and abstraction for better design.

#### 3. Liskov's Substitution Principle (LSP)
_<p align="center">Objects of a subclass must be usable in place of objects of their superclass without altering the program’s correctness</p>_
##### Example of violating LSP

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def set_width(self, width):
        self.width = width
    def set_height(self, height):
        self.height = height
    def area(self):
        return self.width * self.height
class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)
    def set_width(self, width):
        self.width = self.height = width
    def set_height(self, height):
        self.width = self.height = height
````

From a programming point of view, the `Square` class may be defined as extending the `Rectangle` class.<br>
However, this violates LSP even though the _is-a_ relationship between `Rectangle` and `Square`
> `Square` is-a `Rectangle`

Consider the below function `resize_to_5_by_4` when a square is passed in, and the open-closed principle might be considered to have been violated.

````python

def resize_to_5_by_4(rectangle):
    rectangle.set_width(5)
    rectangle.set_height(4)
    assert rectangle.area() == 20

resize_to_5_by_4(Rectangle(1, 1))  # passes
````

````python
resize_to_5_by_4(Square(1))        # fails: area is 16
````

If it is considered that the type of a shape should only be a constraint on the relationship of its dimensions, then the assumption in `resize_to_5_by_4()` that `setHeight()` will change the height and area, but the width is not valid.<br>
This assumption is invalid not only for squares but potentially other rectangles that might be coded to preserve area or aspect ratio when height changes.

##### Refactoring to follow LSP

````python
from typing import Protocol

class Shape(Protocol):
    def area(self) -> float: ...
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height        
class Square:
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side * self.side
def total_area(shapes: list[Shape]) -> float:
    return sum(shape.area() for shape in shapes)

print(total_area([Rectangle(4,5)]))
print(total_area([Square(4)]))
````

Now both types can be used whenever a `Shape` is expected, without claiming that a square supports rectangle-specific resizing behaviour.

#### 4. Interface Segregation Principle (ISP)
The origins of the principle leads back to the XeroX printer.<br>
It could perform a number of tasks including stapling and faxing.<br>
However as the functionality grew, the software grew as well, increasing the development cycle.

The design problem was that a single `Job` class was used by the majority of tasks.<br>
This resulted in a _fat_ class with multitudes of methods specfic to a varienty of different clients.<br>
Because of this design, a staple job would know all the methods of the print job, even though there was not use of them.

The solution was Interface Segregation Principle.<br>
Applied to the Xerox software, an interface layer between the `Job` class and its clients was added using the [Dependency Inversion Principle](#5-dependency-inversion-principle).

Instead of having one large `Job` class, a `Staple Job` interface and a `Print Job` interace was created to be used by `Staple` and `Print` classes, calling methods of the `Job` class.

##### Violation of ISP

````python
class MultiFunctionDevice(ABC):
    @abstractmethod
    def print(self, text: str): ...
    @abstractmethod
    def scan(self) -> str: ...
    @abstractmethod
    def fax(self, number: str): ...
class SimplePrinter(MultiFunctionDevice):
    def print(self, text: str):
        print(text)
    def scan(self):
        raise NotImplementedError("SimplePrinter does not support scanning")
    def fax(self, number: str):
        raise NotImplementedError("SimplePrinter does not support fax")
SimplePrinter().print('Hello')
    
````

````python
SimplePrinter().scan()
````

Like in the original case, the functionality of `MultiFunctionDevice` includes the ability to print, scan, and fax.<br>
However `SimplePrinter` is forced to implement these methods despite not having the necessary capabilities.

##### Refactoring to follow ISP

````python
class Printer(ABC):
    @abstractmethod
    def print(self, text: str): ...
class Scanner(ABC):
    @abstractmethod
    def scan(self) -> str: ...
class Faxer(ABC):
    @abstractmethod
    def fax(self, number: str): ...

class SimplePrinter(Printer):
    def print(self, text: str):
        print(text)
class AdvancedPrinter(Printer, Scanner):
    def print(self, text: str):
        print(f"[Advanced Print] {text}")
    def scan(self) -> str:
        return "Scan complete"
class OfficeMachine(Printer, Scanner, Faxer):
    def print(self, text: str):
        print("[Office Device Print]", text)
    def scan(self):
        return "[Scan Data]"
    def fax(self, number: str):
        print(f"Fax sent to {number}")

SimplePrinter().print("Hello")
AdvancedPrinter().print("Hello")
print(AdvancedPrinter().scan())
OfficeMachine().print("Hello")
print(OfficeMachine().fax("0123456789"))
````

The advantages of splitting the interface into independent capability interfaces:
* Class responsibilities are clear, interfaces are concise
* Users only depend on the interfaces they need
* Extensions are more flexible, supporting on-demand combinations
* Reduces code redundancy and duplicate implementations
* Testing is easier, focusing only on relevant functionalities
> [Taken from here](https://boardor.com/blog/python-interface-segregation-principle-isp)


#### 5. Dependency Inversion Principle
The principle emphasizes reducing coupling between high-level and low-level modules by introducting abstractions.<br>
The principle states:
> High-level modules should not depend on low-level modules.<br>Both should depend on abstractions.

> Abstractions should not depend on details.<br>Details should depend on abstractions.

##### Violation of DIP

````python
class FrontEnd:
    def __init__(self, back_end):
        self.back_end = back_end
    def display_data(self):
        data = self.back_end.get_data_from_database()
        print("Display data:", data)
class BackEnd:
    def get_data_from_database(self):
        return "Data from the database"
    def get_data_from_api(self):
        return "Data from the API"
````

The `FrontEnd` class displays data to users, using data from a database.

If the data needed to be retrieved from an API, it requires changes to the `BackEnd` because it currently is not supported.<br>
It also requires changes to the `FrontEnd` to display the data from the API.<br>
The current implementation means changes/updates are longer than they should be.

##### Refactoring to follow DIP

````python
class FrontEnd:
    def __init__(self, data_source):
        self.data_source = data_source
    def display_data(self):
        data = self.data_source.get_data()
        print("Display data:", data)
class DataSource(ABC):
    @abstractmethod
    def get_data(self):
        pass
class Database(DataSource):
    def get_data(self):
        return "Data from the database"
class API(DataSource):
    def get_data(self):
        return "Data from the API"
````

By introducing middleware `DataSource` class, the `FrontEnd` uses `DataSource` to provide the data.<br>
Whether the data is pulled from `Database` or `API`, it will be feed into `DataSource` and passed onto `FrontEnd` to be displayed.

DIP helps organise code so that the main parts don't have to change every time small parts does.<br>
Making code more flexible and easier to maintain.

## Web/API Basics
### 1. What is an API
An API (Application Programming Interface) allows software to communicate with each other.

A restaurant example:
* Customer &rarr; Client

* Kitchen &rarr; Server

Customer tells the waiter what they want.<br>
&emsp;&emsp;The waiter takes the request to the kitchen.<br>
&emsp;&emsp;&emsp;&emsp;The kitchen brings back the response.
```
Client --> API --> Server
Client <-- API <-- Server
```

* Log in to Spotify.

* Browse Amazon products

### APIRouter (organizing larger applications)
As applications grow, keeping all routes in onefile becomes messy.

Bad:
```py

```



Benefits:
* Cleaner code
* Better organisation
* Easier maintenance
