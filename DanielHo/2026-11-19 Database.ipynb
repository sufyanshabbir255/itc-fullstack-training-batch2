# Databases

## 1. SQL vs. NoSQL
### Structured Query Language (Relational Databases) 
Data is stored in tables.
```
User table
 _______________________
| id | Name       | Age |
|----|------------|-----|
| 1  | Alice      | 30  |
| 2  | Bob        | 25  |
| 3  | Charlie    | 35  |


Orders Table
 __________________________
| id | user_id | product_id |
|----|---------|------------|
| 1  | 1       |         3  |
| 2  | 2       |         2  |
| 3  | 2       |         3  |
```

Relationships are created using keys.

### Not Only SQL (Non-relational)
Data is commonly stored as documents
```json
{
     "id"       : 1    
    ,{name}     : "John"
    ,{orders}   : [
        {
            "product"   : "Laptop"
        }
    ]
}
```

Everything can sit in one document

### Comparison
| SQL | NoSQL |
| :-- | :-- |
| Tables | Documents |
| Structured schema | Flexible schema |
| Relationships | Embedded data |
| ACID transactions | Usually prioritises scalabilty |
| Complex queries | Fast flexible storage |
| Best for business systems | Best for large distributed systems |##

## 2. Relationship Types
### One-to-One (1:1)
```
 User table
 _________________
| id | Name       |
|----|------------|
| 1  | Alice      | 


Profile Table
 __________________________
| id | user_id | bio       |
|----|---------|-----------|
| 1  |       1 | Developer |
```

One user    ⇒ One profile<br>
One profile ⇒ One user

### One-to-Many (1:N)
One user can have many orders
```
User table
 _________________
| id | Name       |
|----|------------|
| 2  | Bob        |


Order table
 ______________
| id | user_id |
|----|---------|
| 1  |       2 |
| 2  |       2 |
| 3  |       2 |
```

One user    ⇒ Many orders<br>
Each order  ⇒ One user


### Many-to-Many (N:N)
One student can have many courses.<br>
One course can have many students.

```
Student A → Python
Student A → SQL

Student B → Python
Student B → JavaScript
```

We need to use a junction table

```
Students                Courses
 _________________       _________________
| id | name       |     | id | name       |
|----|------------|     |----|------------|
| 1  | Alice      |     | 1  | Python     |
| 2  | Bob        |     | 2  | SQL        |
                        | 3  | JavaScript |


StudentCourse
 ________________________
| student_id | course_id |
|------------|-----------|
| 1          | 1         |
| 1          | 2         |
| 2          | 1         |
| 2          | 3         |
```

This links both tables

## 3. Primary key vs. Foreign key
### Primary Key (PK)
Uniquely identifies a row
```
User table
 _________________
| id | Name       |
|----|------------|
| 1  | Alice      |
| 2  | Bob        |
```
`id` is the primary key.

Requirements:
- Unique
- Not null

### Foreign Key (FK)
References another table's primary key.
```
Order table
 ______________
| id | user_id |
|----|---------|
| 1  |       2 |
| 2  |       2 |
```
`user_id` is a foreign key.<br>
It points to `id` found in the `user` table (`user.id`).

## 4. Indexing
Without an index, the database scans every row.

Imagine a table with 1,000,000 rows and looking for the 999,999th row.

An index jumps to the exact row by creating a searchable structure.

## 5. Why indexing has diminishing returns
Whilst read performance improves, write performance decreases.<br>
As the data changes, the index must also be updated.<br>
Index also consumes spaces.

Best practice is to index frequently searched columns:
- email
- username
- foreign keys
- order_name
