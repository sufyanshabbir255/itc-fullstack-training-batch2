## FizzBuzz
print “FizzBuzz” for multiples of 15, “Fizz” for 3, “Buzz” for 5, otherwise the number.

```python
def fizz_buzz(n):
    result = []
    for i in range(1, n+1):
        if i % 3 == 0 and i % 5 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result

input_number = int(input("Enter a number to FizzBuzz to: "))
fizz_buzz_result = fizz_buzz(input_number)
print("FizzBuzz result:", fizz_buzz_result)
```

## Factorial Finder 
The Factorial of a positive integer, n, is defined as the product of the sequence n, n-1, n-2, ...1 and the factorial of zero, 0, is defined as being 1. 

Solve this using both loops and recursion.

```python
def factoiral_finder(n):
    if n == 0:
        return 1
    else:
        return n * factoiral_finder(n - 1)
    
factorial_number = input("Give a number to find the factorial of: ")
factorial_result = factoiral_finder(int(factorial_number))
print(f"The factorial of {factorial_number} is: {factorial_result}")
```

## Fibbing
Create a program that will calculate the Fibonacci Sequence to 10 places.

Extensions:
1. Allow the user to specify the number of places generated
2. Print this in reverse order
3. Display the total of all the numbers shown

```python
def fibonacci_sequencing(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_sequence = [0, 1]
        for i in range(2, n):
            next_number = fib_sequence[i - 1] + fib_sequence[i - 2]
            fib_sequence.append(next_number)
        return fib_sequence

fib_number = input("Enter a number to generate Fibonacci sequence: ")
fib_sequence_result = fibonacci_sequencing(int(fib_number))
print(f"Fibonacci sequence up to {fib_number} terms: {fib_sequence_result}")
print(f'Numbers in reverse order: {fib_sequence_result[::-1]}')
print(f'Sum of the sequence: {sum(fib_sequence_result)}')
```

## Fruit Machine 
Write a program to simulate a Fruit Machine that displays three symbols at random from Cherry, Bell, Lemon, Orange, Star, Skull.

The player starts with £1 credit, with each go costing 20p.<br>
If the Fruit Machine “rolls” two of the same symbol, the user wins 50p.<br>
The player wins £1 for three of the same and £5 for 3 Bells.<br>
The player loses £1 if two skulls are rolled and all of his/her money if three skulls are rolled.<br>
The player can choose to quit with the winnings after each roll or keep playing until there is no money left.

```python
from random import randint

def show_roll(roll):
    result_dict = {
         1: '🍒'
        ,2: '🔔'
        ,3: '🍋'
        ,4: '🍊'
        ,5: '⭐'
        ,6: '💀'
    }
    return f'You rolled: {result_dict[roll[0]]} {result_dict[roll[1]]} {result_dict[roll[2]]}.'

credit = 100

roll = (randint(1, 6), randint(1, 6), randint(1, 6))
print(show_roll(roll))
roll = sorted(roll)

if len(set(roll)) == 1:
    if set(roll) == {6}:
        credit = 0
        print(f'You lost all your credits.')
    elif set(roll) == {2}:
        credit += 500
        print(f'You won 500 credits!\nNew balance: {credit}')
    else:
        credit += 100
        print(f'You won 100 credits!\nNew balance: {credit}')

if roll[0] == roll[1] or roll[1] == roll[2]:
    if roll[1] == 6:
        credit -= 100
        print(f'You lost 100 credits!\nNew balance: {credit}')
    else:
        credit += 50
        print(f'You won 50 credits!\nNew balance: {credit}')
```
