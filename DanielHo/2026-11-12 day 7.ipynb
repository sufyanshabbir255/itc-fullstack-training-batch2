# Day 7
## OAuth2PasswordBearer
`OAuth2PasswordBearer` is a FastAPI security dependency that:
1. Reads the `Authorization` header
2. Extracts the Bearer token.
3. Makes the token available in your route/dependency

```
Authorization Header
        ↓
OAuth2PasswordBearer
        ↓
Extract Token
        ↓
Route
```


## OAuth2PasswordRequestForm
A dependency that extracts login credentials from a form.

OAuth2PassworRequestForm wraps OAuth2 login requirements into one dependency.

## Scopes
Scopes define what actions a token is allowed to perform.



## Role-Based Access Control
Common roles:
* Admin
* Manager
* Editor
* User
* Guest

Differnt roles provide different levels of access.

## Roles vs. Scopes
Roles provide broad permissions to users.

Scope provide specific permissions.
