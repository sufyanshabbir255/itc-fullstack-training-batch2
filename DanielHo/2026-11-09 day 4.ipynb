# Routes
A route connects a URL to a Python function.
Python
```py
from fastapi import FastAPI
app = FastAPI()
@app.get("/")
def home():
    return {"message": "Hello world!"}
```

HTTP
```
GET /
```

Response
```json
{
    "message": "Hello world!"
}

## Common route types
Python
```py
@app.post("/items")
def create_items():
    pass

@app.get("/items")
def get_items():
    pass

@app.out("/items")
def update_items():
    pass

@app.delete("/items{id}")
def update_items(id: ind):
    pass
```

Think:
| Method    | Purpose   |
| :--       | :--       |
| POST      | Create    |

| PUT       | Update    |
| DELETE    | Delete    |


# Path Parameters
Variables inside the URL

Python:
```py
@app.get("/users/{user_id}")
def get_user(user_id: int):
    return {"user_id": user_id}
```

Request:
```
GET /users/7
```

Response:
```json
{
    "user_id"   : 7
}
```


# Request Body
Used mainly with POST and PUT.

Python
```py
from pydantic import BaseModel
class User(Basemodel):
    name    : str
    age     : int

@app.post("/users")
def create_user(user: User):
    return user
```

Request:
```json
{
    "name"  : "John",
    "age"   : 25
}

FastAPI converts JSON into a Python object:
```py
user.name
user.age
```


# Nested Body Models
Real APIs often contain objects inside objects.

```py
from pydantic import BaseModel
class Address(BaseModel):
    city    : str
    country : str

class User(BaseModel):
    name    : str
    address : Address
```

Route:
```py
@app.post("/users")
def create_user(user User):
    return user
```

Request:
```json
{
    "name"      : "John",
    "address"   :  {
        "city"      : "London",
        "country"   : "UK"
    }
}
```

Access values:
```py
user.address.city
```


# Field Constraints
Validation rules
```py
from pydantic import BaseModel, Field
class User(BaseModel):
    username    : str   = Field(
        min_length  = 3,        # Minimum length: 3 characters
        max_length  = 20        # Maximum length: 20 characters
    )
    age         : int   = Field(
        ge          = 18        # Greater or Equal than 18
        lt          = 120       # Less Than 120
    )
    email       : str   = Field(
        pattern = r".+@.+\.com$" # .+ any character at least once
                                 # @ character
                                 # .+ any character at least once
                                 # \.com literal period character and literal 'com'
                                 # $ end of line
    )
```



# Header Parameters
Headers contain metadata

Request
```http
Authorization:  Bearer xyz
User-Agent:     Chrome
```

Route
```py
from fastapi import Header
@app.get("/profile")
def profile(user_agent: str = Header()):
    return {"browser": user_agent}
```
FastAPI reads the header automatically.

# Cookie Parameters
Cookies are stored by the browser.

```py
from fastapi import Cookie
@app.get("/cart")
def cart(session_id: str = Cookie()):
    return {"session_id": session_id}
```

Request:
```http
Cookie: session_id=abc123
```

Response:
```json
{
    "session_id": "abc123"
}
```

# Form Parameters
HTML forms don't send JSON
```py
from fastapi import Form
@app.post("/login")
def login(
    username: str = Form(),
    password: str = Form()
):
    return {"username": username}
```

HTML form sends:
```
username=john
password=secret
```

Accept files from user.
```py
from fastapi import UploadFile, File
@app.post("/upload")
def upload(file: UploadFile = File()):
    return {"filename": file.filename}
```

Upload:
```
resume.pdf
```

Response:
```json
{
    "filename": "resume.pdf"
}
