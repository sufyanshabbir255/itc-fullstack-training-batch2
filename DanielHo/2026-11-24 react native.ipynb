Instead of using Kotlin to write Android applications and Swift for iPhone applications, React Native can be use for both.


## State
State stores data.
```jsx
const [count, setCount] = useState(0);
```

Example
```jsx
import React, { useState } from "react";
import { View, Text, Button } from "react-native";

export default function App() {
    const [count, setCount] = useState(0);

    return (
        <View>
            <Text>{count}</Text>

            <Button
                title="Increment"
                onPress={() => setCount(count + 1)}
            />
        </View>
    );
}
```

## Handling User Input
```jsx
const [email, setEmail] = useState("");
```

Input:
```jsx
<TextInput
    value           = {email}
    onChangeText    = {setEmail}
/>
```

When user types `jason@test.com`, state updates automatically.

## Calling FastAPI
In FastAPI, we get a list of users
```py
@app.get("/users")
async def get_users():
    return [...]
```

In React Native we call and wait for the response data.
```jsx
const getUsers = async () => {
    const response = await fetch (
        "http://localhost:8000/users"
    );

    const data = await response.json();

    console.log(data);
}

## useEffect
Runs code when component loads
```jsx
useEffect(() => {
    getUsers();
}, []);
```

Screen opens → `useEffect` runs → API called → Users shown
